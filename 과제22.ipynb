quotes 크롤링 (1~10페이지)

```python
import requests
from bs4 import BeautifulSoup

quotes_list = []

for page in range(1, 11):
    url = f"http://quotes.toscrape.com/page/{page}/"
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    quotes = soup.select(".quote .text")

    for q in quotes:
        quotes_list.append(q.get_text())

print(len(quotes_list))
quotes_list[:5]
```

CountVectorizer 적용

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(stop_words='english')

X = vectorizer.fit_transform(quotes_list)

feature_names = vectorizer.get_feature_names_out()
```

. 가장 많이 나온 단어 TOP 5

```python
import numpy as np

word_counts = X.sum(axis=0)
word_counts = np.array(word_counts).flatten()

top_indices = word_counts.argsort()[::-1][:5]

for i in top_indices:
    print(feature_names[i], word_counts[i])
```

0/1 Encoding 데이터프레임 만들기

```python
import pandas as pd

binary_vectorizer = CountVectorizer(stop_words='english', binary=True)

X_binary = binary_vectorizer.fit_transform(quotes_list)

df = pd.DataFrame(X_binary.toarray(), columns=binary_vectorizer.get_feature_names_out())

df.head()
```

```python
import requests
from bs4 import BeautifulSoup

quotes_list = []

for page in range(1, 11):
    url = f"http://quotes.toscrape.com/page/{page}/"
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    quotes = soup.find_all("span", class_="text")

    for q in quotes:
        quotes_list.append(q.get_text())

print(len(quotes_list))
```

```python
import requests
from bs4 import BeautifulSoup

quotes_list = []
page = 1

while True:
    url = f"http://quotes.toscrape.com/page/{page}/"
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    quotes = soup.find_all("span", class_="text")

    if not quotes:
        break

    for q in quotes:
        quotes_list.append(q.get_text())

    page += 1

print(len(quotes_list))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(quotes_list)

df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())

df
```
