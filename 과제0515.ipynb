```python
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = 'all'

import warnings
warnings.filterwarnings(action='ignore')

import pandas as pd
import numpy as np

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

import gradio as gr
```

```python
wine = load_wine()

X = pd.DataFrame(
    wine.data,
    columns=wine.feature_names
)

y = pd.Series(
    wine.target,
    name='target'
)

df = pd.concat(
    [X, y],
    axis=1
)

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.dropna()

df = df.drop_duplicates()

df.shape
```

```python
df['target'].value_counts()
```

```python
df['target'].value_counts(normalize=True)
```

```python
data = df[[
    'alcohol',
    'flavanoids',
    'color_intensity'
]]

target = df['target']

data.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    data,
    target,
    test_size=0.2,
    random_state=0,
    stratify=target
)
```

```python
model = RandomForestClassifier(
    n_estimators=300,
    max_depth=5,
    random_state=0
)

model.fit(
    X_train,
    y_train
)
```

```python
pred = model.predict(X_test)

accuracy_score(
    y_test,
    pred
)
```

```python
confusion_matrix(
    y_test,
    pred
)
```

```python
confusion_matrix(
    y_test,
    pred
)
```

```python
print(
    classification_report(
        y_test,
        pred,
        target_names=wine.target_names
    )
)
```

```python
importance = pd.DataFrame({
    'feature': data.columns,
    'importance': model.feature_importances_
})

importance.sort_values(
    by='importance',
    ascending=False
)
```

```python
def wine_predict(
    alcohol,
    flavanoids,
    color_intensity
):

    input_data = pd.DataFrame(
        [[
            alcohol,
            flavanoids,
            color_intensity
        ]],
        columns=[
            'alcohol',
            'flavanoids',
            'color_intensity'
        ]
    )

    pred = model.predict(input_data)[0]

    proba = model.predict_proba(input_data)[0]

    result = wine.target_names[pred]

    probability = pd.DataFrame({
        'Wine Class': wine.target_names,
        'Probability': proba
    })

    return result, probability
```

```python
app = gr.Interface(
    fn=wine_predict,

    inputs=[
        gr.Number(
            label='Alcohol',
            value=13.0
        ),

        gr.Number(
            label='Flavanoids',
            value=2.5
        ),

        gr.Number(
            label='Color Intensity',
            value=5.0
        )
    ],

    outputs=[
        gr.Textbox(
            label='Predicted Wine Class'
        ),

        gr.Dataframe(
            label='Prediction Probability'
        )
    ],

    title='Wine Classification Prediction Service',

    description='''
    load_wine 데이터셋을 이용한 와인 분류 예측 서비스입니다.
    3개의 특성변수(alcohol, flavanoids, color_intensity)를 사용하여
    RandomForestClassifier 모델로 와인의 클래스를 예측합니다.
    '''
)

app.launch(
    share=True
)
```
