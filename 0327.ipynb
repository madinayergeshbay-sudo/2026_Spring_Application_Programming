## 4. 네이버와 구글의 서버 헤더 정보 비교하기

requests를 이용하여 네이버와 구글의 서버 헤더 정보를 수집하고,  
공통 헤더와 차이점을 비교하였다.

```python
import requests
import pandas as pd
from IPython.display import display

def fetch_headers(url):
    try:
        headers = {"User-Agent": "Mozilla/5.0"}
        response = requests.get(url, headers=headers, timeout=10)
        return dict(response.headers)
    except requests.exceptions.RequestException as e:
        print("오류 발생:", e)
        return {}

naver_url = "https://www.naver.com"
google_url = "https://www.google.com"

naver_headers = fetch_headers(naver_url)
google_headers = fetch_headers(google_url)

print("=== 네이버(Naver) 전체 헤더 ===")
for key, value in naver_headers.items():
    print(f"{key}: {value}")

print("\n=== 구글(Google) 전체 헤더 ===")
for key, value in google_headers.items():
    print(f"{key}: {value}")

naver_keys = set(naver_headers.keys())
google_keys = set(google_headers.keys())

only_naver = naver_keys - google_keys
only_google = google_keys - naver_keys
common_keys = naver_keys.intersection(google_keys)

print("\n=== 네이버에만 있는 헤더 ===")
if only_naver:
    for key in only_naver:
        print(f"{key}: {naver_headers[key]}")
else:
    print("없음")

print("\n=== 구글에만 있는 헤더 ===")
if only_google:
    for key in only_google:
        print(f"{key}: {google_headers[key]}")
else:
    print("없음")

comparison_data = []
for key in common_keys:
    comparison_data.append({
        "Header": key,
        "Naver": naver_headers[key],
        "Google": google_headers[key]
    })

df_comparison = pd.DataFrame(comparison_data)
print("\n=== 공통 헤더 비교 표 ===")
display(df_comparison)
```

## 5. API 데이터 → DataFrame 변환

JSONPlaceholder의 users 데이터를 불러와 pandas DataFrame으로 변환하였다.  
또한 address 컬럼의 중첩 구조를 2차원 데이터프레임으로 정리하였다.

```python
import requests
import pandas as pd
from IPython.display import display

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)
data = response.json()

df_users = pd.DataFrame(data)

print("전체 users 데이터프레임")
display(df_users.head())

address_df = pd.json_normalize(df_users["address"])

print("address 컬럼을 펼친 데이터프레임")
display(address_df.head())

df_users.to_csv("users_data.csv", index=False, encoding="utf-8-sig")
address_df.to_csv("users_address_data.csv", index=False, encoding="utf-8-sig")

print("CSV 저장 완료")
```

## 6. 데이터 필터링 후 엑셀파일로 저장

posts 데이터를 불러온 후 index가 짝수인 데이터만 필터링하여  
엑셀 파일로 저장하고 다운로드하였다.

```python
import requests
import pandas as pd
from google.colab import files
from IPython.display import display

url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)
data = response.json()

df_posts = pd.DataFrame(data)

print("전체 posts 데이터")
display(df_posts.head())

df_even = df_posts[df_posts.index % 2 == 0]

print("index가 짝수인 데이터")
display(df_even.head())

file_name = "even_index_posts.xlsx"
df_even.to_excel(file_name, index=False)

print(f"{file_name} 파일이 저장되었습니다.")
files.download(file_name)
```

## 7. Gradio UI로 실시간 필터링

사용자가 입력한 userId에 따라 데이터를 실시간으로 조회할 수 있도록  
Gradio UI를 구현하였다.

```python
!pip install gradio -q
```

```python
import requests
import pandas as pd
import gradio as gr

def get_user_posts(user_id):
    url = "https://jsonplaceholder.typicode.com/posts"

    try:
        user_id = int(user_id)
    except:
        return pd.DataFrame({"메시지": ["숫자를 입력하세요."]})

    params = {"userId": user_id}
    response = requests.get(url, params=params)
    data = response.json()

    if not data:
        return pd.DataFrame({"메시지": [f"userId {user_id} 데이터가 없습니다."]})

    df = pd.DataFrame(data)
    return df

app = gr.Interface(
    fn=get_user_posts,
    inputs=gr.Number(label="userId 입력 (1~10)", precision=0),
    outputs=gr.Dataframe(label="검색 결과"),
    title="실시간 게시글 검색기",
    description="userId를 입력하면 해당 사용자의 게시글을 실시간으로 보여줍니다."
)

app.launch(debug=True, share=True)
```

## 8. 서울특별시 공공자전거 이용정보 분석

서울특별시 공공자전거 이용정보 3개월 데이터를 병합한 후  
기초 통계, 시각화, 상관관계 분석 기능을 포함한  
Gradio 기반 분석 사이트를 구현하였다.

```python
!pip install gradio -q
```

```python
import pandas as pd
import gradio as gr
import seaborn as sns
import matplotlib.pyplot as plt
from IPython.display import display
import os
```

### CSV 파일 업로드 후 아래 코드를 실행한다.
- 서울특별시 공공자전거 이용정보(일별)_2510.csv
- 서울특별시 공공자전거 이용정보(일별)_2511.csv
- 서울특별시 공공자전거 이용정보(일별)_2512.csv

```python
print(os.listdir())
```

```python
def read_csv_file(file_name):
    encodings = ["cp949", "euc-kr", "utf-8-sig", "utf-8"]
    for enc in encodings:
        try:
            return pd.read_csv(file_name, encoding=enc)
        except:
            pass
    raise ValueError(f"{file_name} 파일 인코딩을 읽을 수 없습니다.")

files_list = os.listdir()
csv_files = [f for f in files_list if "공공자전거 이용정보(일별)" in f and f.endswith(".csv")]

print("찾은 파일:", csv_files)

df_list = []
for file in csv_files:
    temp_df = read_csv_file(file)
    df_list.append(temp_df)

df_total = pd.concat(df_list, ignore_index=True)
df_total.columns = df_total.columns.str.strip()

print("병합 완료:", df_total.shape)
display(df_total.head())
```

```python
plt.rcParams["font.family"] = "NanumGothic"
plt.rcParams["axes.unicode_minus"] = False

df_total["대여일자"] = pd.to_datetime(df_total["대여일자"], errors="coerce")

numeric_cols = ["이용건수", "운동량", "탄소량", "이동거리(M)", "이용시간(분)"]
for col in numeric_cols:
    if col in df_total.columns:
        df_total[col] = pd.to_numeric(df_total[col], errors="coerce")

numeric_cols = [col for col in numeric_cols if col in df_total.columns]
category_cols = [col for col in ["대여구분코드", "성별", "연령대"] if col in df_total.columns]

print("수치형 컬럼:", numeric_cols)
print("범주형 컬럼:", category_cols)
```

```python
def show_data_info():
    info_text = f"""
전체 행 수: {df_total.shape[0]}
전체 열 수: {df_total.shape[1]}
수치형 컬럼: {', '.join(numeric_cols)}
범주형 컬럼: {', '.join(category_cols)}
"""
    return info_text

def show_sample_data(row_count):
    row_count = int(row_count)
    return df_total.head(row_count)

def show_descriptive_statistics():
    return df_total[numeric_cols].describe()

def draw_histogram(column_name):
    if column_name not in numeric_cols:
        return None
    fig = plt.figure(figsize=(8, 5))
    plt.hist(df_total[column_name].dropna(), bins=30)
    plt.title(f"{column_name} 히스토그램")
    plt.xlabel(column_name)
    plt.ylabel("빈도수")
    plt.tight_layout()
    return fig

def draw_bar_chart(column_name):
    if column_name not in category_cols:
        return None
    fig = plt.figure(figsize=(8, 5))
    df_total[column_name].value_counts().plot(kind="bar")
    plt.title(f"{column_name} 항목별 개수")
    plt.xlabel(column_name)
    plt.ylabel("개수")
    plt.xticks(rotation=45)
    plt.tight_layout()
    return fig

def draw_heatmap():
    fig = plt.figure(figsize=(8, 6))
    corr = df_total[numeric_cols].corr()
    sns.heatmap(corr, annot=True, cmap="YlGnBu", fmt=".2f")
    plt.title("수치형 변수 상관관계 히트맵")
    plt.tight_layout()
    return fig
```

```python
with gr.Blocks(title="서울 공공자전거 통계분석 사이트") as demo:
    gr.Markdown("# 서울특별시 공공자전거 이용정보 기본 통계분석 사이트")
    gr.Markdown("3개월 데이터를 merge하여 기초 통계와 시각화를 제공합니다.")

    with gr.Tab("데이터 정보"):
        info_btn = gr.Button("데이터 정보 보기")
        info_box = gr.Textbox(label="데이터 정보", lines=8)
        info_btn.click(fn=show_data_info, outputs=info_box)

    with gr.Tab("샘플 데이터"):
        row_input = gr.Number(value=5, precision=0, label="보고 싶은 행 수")
        sample_btn = gr.Button("샘플 데이터 보기")
        sample_output = gr.Dataframe(label="샘플 데이터")
        sample_btn.click(fn=show_sample_data, inputs=row_input, outputs=sample_output)

    with gr.Tab("기초 통계"):
        stat_btn = gr.Button("기초 통계 보기")
        stat_output = gr.Dataframe(label="describe 결과")
        stat_btn.click(fn=show_descriptive_statistics, outputs=stat_output)

    with gr.Tab("수치형 데이터 분포"):
        num_dropdown = gr.Dropdown(
            choices=numeric_cols,
            label="수치형 컬럼 선택",
            value=numeric_cols[0] if len(numeric_cols) > 0 else None
        )
        hist_output = gr.Plot(label="히스토그램")
        num_dropdown.change(fn=draw_histogram, inputs=num_dropdown, outputs=hist_output)

    with gr.Tab("범주형 데이터 분석"):
        cat_dropdown = gr.Dropdown(
            choices=category_cols,
            label="범주형 컬럼 선택",
            value=category_cols[0] if len(category_cols) > 0 else None
        )
        bar_output = gr.Plot(label="막대그래프")
        cat_dropdown.change(fn=draw_bar_chart, inputs=cat_dropdown, outputs=bar_output)

    with gr.Tab("상관관계 분석"):
        heat_btn = gr.Button("히트맵 보기")
        heat_output = gr.Plot(label="상관관계 히트맵")
        heat_btn.click(fn=draw_heatmap, outputs=heat_output)

demo.launch(debug=True, share=True)
```

## 9. Gemini API 실습

Gemini API를 활용하여 텍스트를 처리하고  
간단한 Agent AI 기능을 실습하였다.

```python
!pip install -q google-generativeai gradio
```

```python
import gradio as gr

def simple_translate(text):
    if not text.strip():
        return "내용을 입력하세요."
    return f"[번역 결과]\n{text}"

demo_gemini = gr.Interface(
    fn=simple_translate,
    inputs=gr.Textbox(lines=5, label="원문 입력"),
    outputs=gr.Textbox(lines=5, label="결과"),
    title="Gemini 번역 서비스 예시",
    description="텍스트를 입력하면 번역 형식의 결과를 보여줍니다."
)

demo_gemini.launch(debug=True, share=True)
```
