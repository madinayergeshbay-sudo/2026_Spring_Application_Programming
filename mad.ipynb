Web Crawling and Scrapping 실습

기본 연습 — 책 제목 20개 가져오기

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "http://books.toscrape.com/"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

titles = [a["title"] for a in soup.select("h3 a")]

for title in titles:
    print(title)
```

제목을 표로 만들기



```python
df_titles = pd.DataFrame(titles, columns=["Book Title"])
df_titles
```

가격도 같이 가져오기



```python
books = soup.select("article.product_pod")

data = []

for book in books:
    title = book.select_one("h3 a")["title"]
    price = book.select_one("p.price_color").get_text(strip=True)
    data.append([title, price])

df_books = pd.DataFrame(data, columns=["Title", "Price"])
df_books
```

가격 숫자로 바꾸기

```python
df_books['Price'] = df_books['Price'].str.extract(r'(\d+\.\d+)')
df_books['Price'] = df_books['Price'].astype(float)
```

제일 비싼 책 5개 보기

```python
top5 = df_books.sort_values(by="Price", ascending=False).head(5)
top5
```

가격 통계 보기

```python
df_books["Price"].describe()
```

카테고리 이름 가져오기

```python
categories = [a.get_text(strip=True) for a in soup.select("ul.nav-list ul li a")]

for c in categories:
    print(c)
```

카테고리 10개만 가져오기

```python
categories = [a.get_text(strip=True) for a in soup.select("ul.nav-list ul li a")[:10]]

df_categories = pd.DataFrame(categories, columns=["Category"])
df_categories
```

책 제목 + 가격 txt 파일로 저장

```python
with open("books.txt", "w", encoding="utf-8") as f:
    for i in range(len(df_books)):
        f.write(df_books.loc[i, "Title"] + " | " + str(df_books.loc[i, "Price"]) + "\n")
```

여러 페이지 연습

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

all_data = []

for page in range(1, 6):
    url = f"https://books.toscrape.com/catalogue/page-{page}.html"
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    books = soup.select("article.product_pod")

    for book in books:
        title = book.select_one("h3 a")["title"]
        price = book.select_one("p.price_color").get_text(strip=True)
        all_data.append([title, price])

    print(f"{page} page 완료")
    time.sleep(0.5)

df_all = pd.DataFrame(all_data, columns=["Title", "Price"])
df_all
```

여러 페이지 가격 숫자로 바꾸기

```python
df_all['Price'] = df_all['Price'].str.extract(r'(\d+\.\d+)')
df_all['Price'] = df_all['Price'].astype(float)

df_all
```
