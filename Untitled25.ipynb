```python
import gradio as gr
import sqlite3
import pandas as pd
import os

DB_NAME = "customer_data.db"
CSV_URL = "https://raw.githubusercontent.com/ancestor9/data/main/customers.csv"


def get_connection():
    return sqlite3.connect(DB_NAME)


def load_table():
    if not os.path.exists(DB_NAME):
        return pd.DataFrame(columns=["id", "username", "full_name"])

    conn = get_connection()
    df = pd.read_sql_query("SELECT * FROM customers ORDER BY id ASC", conn)
    conn.close()
    return df


def init_and_load(csv_url, top_n):
    try:
        if os.path.exists(DB_NAME):
            os.remove(DB_NAME)

        conn = get_connection()
        cursor = conn.cursor()

        cursor.execute("""
        CREATE TABLE customers (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            username TEXT NOT NULL UNIQUE,
            full_name TEXT
        )
        """)

        df = pd.read_csv(csv_url)
        df_top = df.head(int(top_n))

        data = []
        for _, row in df_top.iterrows():
            customer_id = str(row["고객ID"])
            data.append((customer_id, customer_id))

        cursor.executemany(
            "INSERT INTO customers (username, full_name) VALUES (?, ?)",
            data
        )

        conn.commit()
        conn.close()

        return f"DB 재생성 후 상위 {top_n}건 적재 완료", load_table()

    except Exception as e:
        return f"처리 중 오류 발생: {e}", load_table()


def create_customer(username, full_name):
    try:
        if not username.strip():
            return "고객 ID를 입력하세요.", load_table()

        conn = get_connection()
        cursor = conn.cursor()

        cursor.execute(
            "INSERT INTO customers (username, full_name) VALUES (?, ?)",
            (username, full_name)
        )

        conn.commit()
        conn.close()

        return f"고객 등록 완료 -> {username}", load_table()

    except Exception as e:
        return f"처리 중 오류 발생: {e}", load_table()


def update_customer(username, full_name):
    try:
        if not username.strip():
            return "수정할 고객 ID를 입력하세요.", load_table()

        conn = get_connection()
        cursor = conn.cursor()

        cursor.execute(
            "UPDATE customers SET full_name = ? WHERE username = ?",
            (full_name, username)
        )

        conn.commit()
        conn.close()

        return f"고객 정보 변경 완료 -> {username}", load_table()

    except Exception as e:
        return f"처리 중 오류 발생: {e}", load_table()


def delete_customer(username):
    try:
        if not username.strip():
            return "삭제할 고객 ID를 입력하세요.", load_table()

        conn = get_connection()
        cursor = conn.cursor()

        cursor.execute(
            "DELETE FROM customers WHERE username = ?",
            (username,)
        )

        conn.commit()
        conn.close()

        return f"고객 삭제 처리 완료 -> {username}", load_table()

    except Exception as e:
        return f"처리 중 오류 발생: {e}", load_table()


with gr.Blocks() as demo:
    gr.Markdown("# 06주차 고객 정보 SQLite 관리")
    gr.Markdown("CSV 데이터를 불러와 customers 테이블에 대해 추가, 수정, 삭제를 수행합니다.")

    with gr.Tab("DB 준비"):
        csv_url = gr.Textbox(label="불러올 CSV 주소", value=CSV_URL)
        top_n = gr.Slider(1, 100, value=10, step=1, label="가져올 상위 데이터 수")
        init_btn = gr.Button("DB 초기화 및 데이터 적재")
        init_msg = gr.Textbox(label="처리 결과")

    with gr.Tab("고객 등록"):
        username = gr.Textbox(label="고객 ID")
        fullname = gr.Textbox(label="고객 이름")
        create_btn = gr.Button("신규 고객 추가")
        create_msg = gr.Textbox(label="처리 결과")

    with gr.Tab("고객 수정"):
        up_username = gr.Textbox(label="수정 대상 고객 ID")
        up_fullname = gr.Textbox(label="변경할 이름")
        update_btn = gr.Button("고객 정보 수정")
        update_msg = gr.Textbox(label="처리 결과")

    with gr.Tab("고객 삭제"):
        del_username = gr.Textbox(label="삭제 대상 고객 ID")
        delete_btn = gr.Button("고객 정보 삭제")
        delete_msg = gr.Textbox(label="처리 결과")

    table = gr.Dataframe(label="customers 데이터 조회 결과", interactive=False)

    init_btn.click(
        fn=init_and_load,
        inputs=[csv_url, top_n],
        outputs=[init_msg, table]
    )

    create_btn.click(
        fn=create_customer,
        inputs=[username, fullname],
        outputs=[create_msg, table]
    )

    update_btn.click(
        fn=update_customer,
        inputs=[up_username, up_fullname],
        outputs=[update_msg, table]
    )

    delete_btn.click(
        fn=delete_customer,
        inputs=[del_username],
        outputs=[delete_msg, table]
    )

demo.launch()
```
