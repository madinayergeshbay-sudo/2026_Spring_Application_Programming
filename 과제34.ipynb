```python
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = 'all'

import warnings
warnings.filterwarnings(action='ignore')

import pandas as pd

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

import gradio as gr
```

```python
wine = load_wine()

X = pd.DataFrame(wine.data, columns=wine.feature_names)
y = pd.Series(wine.target)

X.head()
y.value_counts()
```

```python
data = X[['alcohol', 'flavanoids', 'color_intensity']]
data.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    data, y, test_size=0.2, random_state=0
)
```

```python
model = RandomForestClassifier(random_state=0)

model.fit(X_train, y_train)

pred = model.predict(X_test)

accuracy_score(y_test, pred)
```

```python
model = RandomForestClassifier(random_state=0)

model.fit(X_train, y_train)

pred = model.predict(X_test)

accuracy_score(y_test, pred)
```

```python
model = RandomForestClassifier(random_state=0)

model.fit(X_train, y_train)

pred = model.predict(X_test)

accuracy_score(y_test, pred)
```

```python
def wine_predict(alcohol, flavanoids, color_intensity):
    input_data = pd.DataFrame([[alcohol, flavanoids, color_intensity]],
                              columns=['alcohol', 'flavanoids', 'color_intensity'])

    result = model.predict(input_data)[0]
    wine_name = wine.target_names[result]

    return wine_name
```

```python
app = gr.Interface(
    fn=wine_predict,
    inputs=[
        gr.Number(label="alcohol"),
        gr.Number(label="flavanoids"),
        gr.Number(label="color_intensity")
    ],
    outputs=gr.Textbox(label="Wine Class Prediction"),
    title="Wine 예측 서비스",
    description="load_wine 데이터에서 3개의 특성변수를 사용하여 와인 클래스를 예측합니다."
)

app.launch(share=True)
```
